```python
import tensorflow as tf
```

**********this section I practiced n-dimensions with tf.constant integer variables

```python
scale = tf.constant(1)
```

```python
scale
```

```python
scale.ndim
```

```python
vector = tf.constant([1 , 2])
```

```python
vector
```

```python
vector.ndim
```

```python
vector2 = tf.constant([[1 , 2],
                        [3 , 4]])
```

```python
vector2
```

```python
vector2.ndim
```

```python
matrix = tf.constant([[[1 , 2] , [5 , 6]] , [[3 , 4] , [7 , 8]]])
```

```python
matrix
```

```python
matrix.ndim
```

********* as you can see we can make variant tf.constant with different dimension
now its time to change the type and see results

```python
floatsample = tf.constant(1.1)
```

```python
floatsample
```

```python
floatsample.ndim
```

```python
float16sample = tf.constant([1.0 , 2.1], dtype=tf.float16)
```

```python
float16sample
```

```python
float16sample.ndim
```
