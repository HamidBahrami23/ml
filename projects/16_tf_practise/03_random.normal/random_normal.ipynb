creating random with tf.random

```python
import tensorflow as tf
```

```python
branch_1 = tf.random.Generator.from_seed(23)
branch_1 = branch_1.normal(shape = (3 , 4))
branch_2 = tf.random.Generator.from_seed(23)
branch_2 = branch_2.normal(shape = (3 , 4))

branch_1 , branch_2 , branch_1 == branch_2
```
