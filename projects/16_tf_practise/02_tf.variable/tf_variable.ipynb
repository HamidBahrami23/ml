creating tensors with tf.variable

```python
import tensorflow as tf
```

```python
changable_tensor = tf.Variable(1)
```

```python
changable_tensor
```

```python
changable_tensor2 = tf.Variable([1])
```

```python
changable_tensor2
```

```python
changable_tensor3 = tf.Variable([[1, 2, 3 ], [4, 5, 6], [7, 8, 9]])
```

```python
changable_tensor3
```

```python
changable_tensor.assign(23)
```

```python
changable_tensor
```

```python
changable_tensor2[0].assign(17)
```

```python
changable_tensor2
```

```python
changable_tensor3[0].assign([11,22,33])
```

```python
changable_tensor3
```
