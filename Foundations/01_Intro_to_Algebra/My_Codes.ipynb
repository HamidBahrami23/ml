## 3. L Norm



L-norms are a way to measure the magnitude or "length" of a vector. They are commonly used in machine learning for regularization, distance calculations, and to understand the scale of vectors. This notebook will cover common L-norms: L1, L2, Squared L2, Max Norm, and Generalized L-p Norms.

### 1. L2 Norm (Euclidean Norm)

The L2 norm, also known as the Euclidean norm, is the most common measure of vector length. It's the distance from the origin to the point represented by the vector.

**Formula:**

||**x**||₂ = √(∑ᵢ |xᵢ|²)

where:

*   **x** is the vector.
*   xᵢ is the i-th element of the vector.
*   ∑ᵢ means summation over all elements of the vector.

### 2. L1 Norm (Manhattan Norm)

The L1 norm, also known as the Manhattan norm or taxicab norm, represents the sum of the absolute values of the vector's components.

**Formula:**

||**x**||₁ = ∑ᵢ |xᵢ|



### 3. Squared L2 Norm

The squared L2 norm is simply the square of the L2 norm.

**Formula:**

||**x**||₂² = ∑ᵢ |xᵢ|²


### 4. Max Norm (L∞ Norm)

The max norm, also known as the infinity norm, is the largest absolute value of the vector's components.

**Formula:**

||**x**||∞ = maxᵢ |xᵢ|

### 5. Generalized L-p Norm

The L-p norm generalizes the concept of L1 and L2 norms.

**Formula:**

||**x**||ₚ = (∑ᵢ |xᵢ|ᵖ)¹/ₚ

**Summary:**

L-norms provide valuable tools for understanding and regularizing machine learning models.  By understanding these different norms, you can better control the behavior and properties of your models.

```python
x = tf.Variable([5. , 2. , 1.])
```

```python
# 1. L1 Norm
l1_norm = tf.reduce_sum(tf.abs(x))
print(f"L1 Norm: {l1_norm.numpy()}")  # Output: L1 Norm: 8.0

# 2. L2 Norm
l2_norm = tf.norm(x)
print(f"L2 Norm: {l2_norm.numpy()}")  # Output: L2 Norm: 5.477225575051661

# 3. Squared L2 Norm
squared_l2_norm = tf.reduce_sum(tf.square(x))
print(f"Squared L2 Norm: {squared_l2_norm.numpy()}") # Output: Squared L2 Norm: 32.0

# 4. Max Norm
max_norm = tf.reduce_max(tf.abs(x))
print(f"Max Norm: {max_norm.numpy()}") # Output: Max Norm: 5.0
```
