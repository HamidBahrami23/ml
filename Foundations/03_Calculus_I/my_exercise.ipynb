```python
A = 1.0      # Amplitude
zeta = 0.1   # Damping ratio (smaller = more oscillation)
wn = 10.0     # Natural frequency

t = np.linspace(0, 20, 500)  

def damped_oscillation(t, A, zeta, wn):
    return A * np.exp(-zeta * wn * t) * np.cos(wn * t)

y = damped_oscillation(t, A, zeta, wn)

plt.plot(t, y)
plt.xlabel("Time (t)")
plt.ylabel("Amplitude")
plt.title("Damped Oscillation")
plt.grid(True)
plt.show()
```

# Computing Derivatives with Differentiation

### Delta Method

```python
x = np.linspace(-10 , 10 , 1000)
```

```python
def s2_fx(x):
    return (x**2 + 2*x + 7)
```

```python
fix, ax = plt.subplots()
_=ax.plot(x , s2_fx(x))
# plt.plot(0.00001 , sin_fx(0.00001) , )
```

```python
def delta_fx(x , delta):
    return (s2_fx(x + delta) - s2_fx(x)) / delta
```

```python
print(delta_fx(2 , 0.1))
print(delta_fx(2 , 0.01))
print(delta_fx(2 , 0.001))
print(delta_fx(2 , 0.0001))
print(delta_fx(2 , 0.00001))
```

# Autodiff

## 1. Pytorch

```python
import torch
```

```python
x = torch.tensor(5.0)
x.requires_grad_() # contagiously track gradients through forward pass
y = x**2
y.backward() # use autodiff
x.grad
```

## 2. Tensorflow

```python
import tensorflow as tf
```

```python
x = tf.Variable(5.0)
```

```python
with tf.GradientTape() as t:
    t.watch(x) # track forward pass
    y = x**2
```

```python
t.gradient(y, x) # use autodiff
```

# Simple Regression

```python
import torch
import matplotlib.pyplot as plt
```

```python
def regression(my_x, my_m, my_b):
    return my_m*my_x + my_b
```

```python
x = torch.tensor([0, 1, 2, 3, 4, 5, 6, 7. , 9 , 10 , 0.1 , 0.12 , -3 , 1.2 , 1.3])
y = x * 2 + 7 + torch.normal(mean=torch.zeros(len(x)), std=0.2)
```

```python
m = torch.tensor([0.9]).requires_grad_()
b = torch.tensor([0.1]).requires_grad_()
```

```python
def mse(my_yhat, my_y): 
    sigma = torch.sum((my_yhat - my_y)**2)
    return sigma/len(my_y)
```

```python
optimizer = torch.optim.SGD([m, b], lr=0.01)
```

```python
epochs = 1000
for epoch in range(epochs):
    
    optimizer.zero_grad() # Reset gradients to zero; else they accumulate
    
    yhat = regression(x, m, b) # Step 1
    C = mse(yhat, y) # Step 2
    
    C.backward() # Step 3
    optimizer.step() # Step 4
    
    print('Epoch {}, cost {}, m grad {}, b grad {}'.format(epoch, '%.3g' % C.item(), '%.3g' % m.grad.item(), '%.3g' % b.grad.item()))
```

```python
def regression_plot(my_x, my_y, my_m, my_b):
    
    fig, ax = plt.subplots()

    ax.scatter(my_x, my_y)
    
    x_min, x_max = ax.get_xlim()
    y_min = regression(x_min, my_m, my_b).detach().item()
    y_max = regression(x_max, my_m, my_b).detach().item()
    
    ax.set_xlim([x_min, x_max])
    _ = ax.plot([x_min, x_max], [y_min, y_max])
```

```python
regression_plot(x, y, m, b)
```

# Sine Function

```python
# src/main.py
def sin_x(x):
  """
  Calculates the sine of x using the Taylor series expansion.

  Args:
    x: The angle in radians.

  Returns:
    The sine of x.
  """
  x = x % (2 * 3.141592653589793)  # Normalize x to be within 0 to 2pi
  result = x
  term = x
  n = 1
  while abs(term) > 1e-15:
    term = -term * x * x / ((2 * n) * (2 * n + 1))
    result += term
    n += 1
  return result
```
